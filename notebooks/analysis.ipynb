```python
from pyspark.sql.functions import col, sum as F_sum, avg, month, year

fact = spark.read.table("final_project.my_project.gold.fact_final")
```

```python
yearly_trends = fact.groupBy("year").agg(
    F_sum("trip_count").alias("trips"),
    F_sum("total_revenue").alias("revenue"),
    avg("avg_pollution").alias("pollution")
)

display(yearly_trends.orderBy("year"))
```

```python
monthly_trends = fact.withColumn("month", month("date")) \
    .groupBy("year", "month") \
    .agg(
        F_sum("trip_count").alias("trips"),
        avg("avg_pollution").alias("pollution")
    )

display(monthly_trends.orderBy("year", "month"))
```

```python
top_zones = fact.groupBy("PU_Zone").agg(
    F_sum("trip_count").alias("trips")
).orderBy(col("trips").desc())

display(top_zones.limit(10))
```

```python
pollution_zones = fact.groupBy("coord").agg(
    avg("avg_pollution").alias("pollution")
).orderBy(col("pollution").desc())

display(pollution_zones.limit(10))
```
